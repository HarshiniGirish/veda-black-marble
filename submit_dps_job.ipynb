# VEDA Black Marble — register + run DPS job

Earthdata token must be a **MAAP Secret** (not a job/CLI input).

```python
from pathlib import Path
from maap.maap import MAAP

maap = MAAP()
SECRET = "EARTHDATA_TOKEN"
existing = maap.secrets.get_secret(SECRET)
if isinstance(existing, dict) and existing.get("code") == 404:
    raise SystemExit(
        f"Create once with: maap.secrets.add_secret('{SECRET}', '<your-token>')"
    )
print(f"MAAP secret '{SECRET}' is present")
```

```python
CONFIG = Path("maap_dps_algorithm_config.yml")
print(maap.register_algorithm_from_yaml_file(str(CONFIG)))
```

```python
job = maap.submitJob(
    identifier="black-marble-smoke",
    algo_id="veda-black-marble",
    version="main",
    queue="maap-dps-worker-16gb",
    bbox="-122.55,37.69,-122.32,37.81",
    date="2023-06-15",
    config="fast",
    osm_source="overpass",
    wgs84="false",
    basename="san_francisco_lights",
    earthdata_secret_name="EARTHDATA_TOKEN",
)
print(job)
```
